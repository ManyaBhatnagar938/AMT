```python
'''Pattern Block - Mapping the notes to numbers, setting up the scales and the pattern dictionary.
Starting at Scale C4
Sa             C4
Komal Re       C#4
Re             D4
Komal Ga       D#4
Ga             E4
Ma             F4
Tivra Ma       F#4
Pa             G4
Komal Dha      G#4
Dha            A4
Komal Ni       A#4
Ni             B4
Sa             C5

Pattern dictionary has the patterns stored in key value pairs with the pattern indexing following P1, P2... format to avoid confusion with the numerical indexes of the
notes.
'''

import random
Note_Map = {"C4": {"name": "Sa","pitch": "C4","midi": 60},
            "C#4": {"name": "Komal Re","pitch": "C#4","midi": 61},
            "D4": {"name": "Re","pitch": "D4","midi": 62},
            "D#4": {"name": "Komal Ga","pitch": "D#4","midi": 63},
            "E4": {"name": "Ga","pitch": "E4","midi": 64},
            "F4": {"name": "Ma","pitch": "F4","midi": 65},
            "F#4": {"name": "Tivra Ma","pitch": "F#4","midi": 66},
            "G4": {"name": "Pa","pitch": "G4","midi": 67},
            "G#4": {"name": "Komal Dha","pitch": "G#4","midi": 68},
            "A4": {"name": "Dha","pitch": "A4","midi": 69},
            "A#4": {"name": "Komal Ni","pitch": "A#4","midi": 70},
            "B4": {"name": "Ni","pitch": "B4","midi": 71},
            "C5": {"name": "Sa","pitch": "C5","midi": 72}}

Diatonic_Scale = ["C4","D4","E4","F4","G4","A4","B4","C5"]
Chromatic_Scale = ["C4","C#4","D4","D#4","E4","F4","F#4","G4","G#4","A4","A#4","B4","C5"]

# Pattern Dictionary

PATTERNS = {"P1": Diatonic_Scale,"P2": Chromatic_Scale,
    "P3": ["C4", "C#4", "E4", "F#4","G4", "G#4", "A#4", "C5"],
    "P4": ["C4", "E4", "D4", "F4","E4", "G4", "F4", "A4","G4", "B4", "A4", "C5"],
    "P5": ["C4", "F4", "D4", "G4","E4", "A4", "F4", "B4","G4", "C5"],
    "P6": ["C4", "D4", "E4", "E4", "E4","D4", "E4", "F4", "F4", "F4","E4", "F4", "G4", "G4", "G4","F4", "G4", "A4", "A4", "A4",
        "G4", "A4", "B4", "B4", "B4","A4", "B4", "C5", "C5", "C5"],
    "P7": ["C4", "D4", "E4","D4", "E4", "F4","E4", "F4", "G4","F4", "G4", "A4","G4", "A4", "B4","A4", "B4", "C5"],
    "P8": ["C4", "C4", "D4", "E4","D4", "D4", "E4", "F4","E4", "E4", "F4", "G4","F4", "F4", "G4", "A4","G4", "G4", "A4", "B4",
        "A4", "A4", "B4", "C5"]}


def get_note(note_code):

    if note_code not in Note_Map:
        raise ValueError(f"Invalid note: {note_code}")

    return Note_Map[note_code]


def get_pattern(pattern):

    """Returns a predefined pattern.
    Example
    get_pattern("P1")
    """

    if pattern not in PATTERNS:
        raise ValueError(f"Unknown pattern: {pattern}")

    return PATTERNS[pattern]


def midi_to_frequency(midi):

    """Converts MIDI pitch to frequency (Hz)"""

    return 440 * (2 ** ((midi - 69) / 12))
```

```python
import random
import pretty_midi

"""
    Generates a monophonic simulation.

    Parameters
    pattern : str
        Pattern ID (e.g. "P1")
    order : str
        "normal", "reverse", "random"
    duration : str or float
        "fixed", "random" or a numeric duration in seconds
    interval : str or float
        "fixed", "random" or a numeric interval in seconds
    velocity : int
        MIDI velocity (0-127)
    instrument : str
        PrettyMIDI instrument name

    Returns
    pm : PrettyMIDI object
    events : list
        Metadata for every generated note
"""
def generate_simulation(pattern,order="normal",duration="fixed",interval="fixed",velocity=100, instrument="Acoustic Grand Piano"):

      sequence = get_pattern(pattern).copy()

      if order.lower() == "reverse":
          sequence.reverse()

      elif order.lower() == "random":
          random.shuffle(sequence)

      elif order.lower() != "normal":
          raise ValueError("Unknown order type")

      # Create MIDI Object

      pm = pretty_midi.PrettyMIDI()

      piano = pretty_midi.Instrument(
          program=pretty_midi.instrument_name_to_program(instrument)
      )

      current_time = 0.2

      events = []

      #Generating Notes

      for note_code in sequence:

          note = get_note(note_code)

          midi_pitch = note["midi"]

          note_name = note["name"]

          frequency = midi_to_frequency(midi_pitch)


          if duration == "fixed":

              note_duration = 0.5

          elif duration == "random":

              note_duration = random.uniform(0.2, 1.0)

          elif isinstance(duration, (int, float)):

              note_duration = duration

          else:

              raise ValueError("Invalid duration option")


          if interval == "fixed":

              note_gap = 0

          elif interval == "random":

              note_gap = random.uniform(0.05, 0.5)

          elif isinstance(interval, (int, float)):

              note_gap = interval

          else:

              raise ValueError("Invalid interval option")


          start_time = current_time

          end_time = start_time + note_duration

          # MIDI Note

          midi_note = pretty_midi.Note(

              velocity=velocity,

              pitch=midi_pitch,

              start=start_time,

              end=end_time

          )

          piano.notes.append(midi_note)

         #Storing data for csv storing

          events.append({

              "note_name": note_name,

              "note_code": note_code,

              "midi": midi_pitch,

              "frequency": round(frequency, 2),

              "start_time": round(start_time, 4),

              "end_time": round(end_time, 4),

              "duration": round(end_time - start_time, 4),

              "interval_after": round(note_gap, 4),

              "velocity": velocity

          })

          current_time = end_time + note_gap

      pm.instruments.append(piano)

      return pm, events
```

```python
#Export Block
import os
import csv
import soundfile as sf

"""
    Saves a simulation as MIDI, WAV and CSV.

    Parameters
    ----------
    pm : PrettyMIDI object

    events : list
        List returned by generate_simulation()

    simulation_name : str
        Example:
        "simulation1"

    output_folder : str
        Folder where files will be saved.

    sampling_rate : int
        Audio sampling rate.
    """


def export_simulation(
    pm,
    events,
    simulation_name,
    output_folder="Simulations",
    sampling_rate=44100):


    #Output folder
    os.makedirs(output_folder, exist_ok=True)

    midi_path = os.path.join(
        output_folder,
        simulation_name + ".mid"
    )

    pm.write(midi_path)

    # Synthesize and Save WAV

    audio = pm.synthesize(fs=sampling_rate)

    wav_path = os.path.join(
        output_folder,
        simulation_name + ".wav"
    )


    sf.write(
        wav_path,
        audio,
        sampling_rate
    )

    #Saving CSV
    csv_path = os.path.join(
        output_folder,
        simulation_name + ".csv"
    )

    with open(csv_path, "w", newline="") as file:

        writer = csv.writer(file)

        writer.writerow([
            "Note",
            "Note Code",
            "MIDI",
            "Frequency (Hz)",
            "Start Time (s)",
            "End Time (s)",
            "Duration (s)",
            "interval_after (s)",
            "Velocity"
        ])

        for note in events:

            writer.writerow([

                note["note_name"],

                note["note_code"],

                note["midi"],

                note["frequency"],

                note["start_time"],

                note["end_time"],

                note["duration"],

                note["interval_after"],

                note["velocity"]

            ])

    print(f"Simulation saved successfully!")

    print(f"MIDI : {midi_path}")
    print(f"WAV  : {wav_path}")
    print(f"CSV  : {csv_path}")
```

```python
# Make all randomness reproducible
random.seed(42)

# Random patterns
PATTERNS["P9"] = random.sample(Diatonic_Scale, 7)
PATTERNS["P10"] = random.sample(Chromatic_Scale, 7)

simulation_settings = [
    {"suffix": "a", "order": "normal",  "duration": "fixed",  "interval": "fixed"},
    {"suffix": "b", "order": "normal",  "duration": "random", "interval": "fixed"},
    {"suffix": "c", "order": "normal",  "duration": "fixed",  "interval": "random"},
    {"suffix": "d", "order": "normal",  "duration": "random", "interval": "random"},
    {"suffix": "e", "order": "reverse", "duration": "fixed",  "interval": "fixed"},
]

for pattern_num in range(1, 11):

    pattern = f"P{pattern_num}"

    for setting in simulation_settings:

        pm, events = generate_simulation(
            pattern=pattern,
            order=setting["order"],
            duration=setting["duration"],
            interval=setting["interval"],
            velocity=100
        )

        simulation_name = f"simulation{pattern_num}{setting['suffix']}"

        export_simulation(
            pm,
            events,
            simulation_name=simulation_name
        )

print("All 50 simulations generated successfully!")
```
